# MorphoBricks preprocessing

This notebook computes one intensity histogram for every 2D plane of the TIFF images in an
experiment folder and saves them in `plane_histograms.csv`. Later MorphoBricks bricks use this
table to normalise images per condition, repeat or image without reading the pixels again.

**How to use it:** run the cells from top to bottom (Shift+Enter) and follow the steps. Each
step shows what it found in plain language; nothing is written until step 3 succeeds.

**Your images are never changed.** Results go to a separate folder next to the experiment
folder (or one you choose), never inside it.

See the [user guide](../docs/user_guide.md) for the folder layout, the messages and the output files.

````python
from morphobricks_preprocessing.notebook_ui import PreprocessingNotebook

# Optionally paste your experiment folder between the quotes, or type it in step 1.
notebook = PreprocessingNotebook(experiment_folder="")
````

## Step 1 · Check the experiment folder

The experiment folder holds up to three levels of folders, whose names become the labels:

```
experiment/condition/biological repeat/technical repeat/image.tif
```

Levels may be left out at the end; the missing labels are `single`. Press **Check folder**: the
folders and image headers are inspected without reading any pixels. Check the tree of labels and
tick the box when it is right. To change a label, rename the folder and check again.

If an image cannot be read (for example a damaged file), you can exclude it and continue.

````python
notebook.folder_step()
````

## Step 2 · Review the axis roles

Each image dimension needs a role: **X** (width), **Y** (height), **Z** (depth), **C** (channel)
or **T** (time). The suggestion comes from the image metadata, so please check it. **Show
preview** displays one plane with the chosen roles; it should look like a normal 2D image. Then
press **Confirm axis roles**.

````python
notebook.axis_step()
````

## Step 3 · Compute the histograms

The defaults suit most experiments. Running again later only reads images that are new or
changed since the last run; rows of deleted or excluded images are removed. Tick *Recompute every
row* to start from scratch.

A plane that needs more working memory than the limit stops the run with a message; raise the
limit only if the computer has enough memory.

````python
notebook.run_step()
````

## Step 4 · Look at the results

Statistics for each group, pooled from the histograms of all its planes.

````python
notebook.results_step()
````

## Using the table in Python

```python
from morphobricks_preprocessing import PooledHistogram, group_records, percentile_range, read_plane_histograms

records = read_plane_histograms("path/to/plane_histograms.csv")
for labels, planes in group_records(records, "condition").items():
    parameters = percentile_range(PooledHistogram(planes), low=1, high=99.8)
    print(labels, parameters)
```
